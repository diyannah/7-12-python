## Exercise 08.1 (function plotting)

Consider the function 
$$
f(x) = e^{x/10} \sin(\omega_{1}x)\sin(\omega_{0}x)
$$
from $x = -4\pi$ to $x = 4\pi$.

(1) Plot the function when $\omega_{0} = \omega_{1} = 1$. Label the axes.

````python
#create a numpy array with (num_points) equally spaced values between 0 and 4pi, including the end points
num_points = 100
x = np.linspace(-4 * np.pi, 4 * np.pi, num=num_points)

f = np.exp(x / 10) * np.sin(x) * np.sin(x) #function we are plotting
plt.xlabel('$x$') #label the x axis
plt.ylabel('$f(x)$') #label the y axis
plt.xlim(x[0],x[-1]) #use start and end values in x as x limits
#plot and display graph
plt.plot(x,f)
plt.show()
````

(2) Create an interactive plot with sliders for $\omega_{0}$ and $\omega_{1}$, varying from 0 to 2.

````python
#import the interactive widgets module ipywidgets to create an interactive plot
from ipywidgets import interact
num_points = 100
#create a function to encapsulate the plot code with the plot parameters as arguments to the function
def plot(ω0, ω1):
    #creat an array of 100 equally spaced points from -4pi to 4pi
    x = np.linspace(-4*np.pi, 4*np.pi, num=num_points)
    plt.xlabel('$x$') #label the x axis
    plt.ylabel('$f(x)$') #label the y axis
    #plot and display the function
    plt.plot(x, np.exp(x/10) * np.sin(ω0*x) * np.sin(ω1*x))
    plt.show()


#use interact function, passing it to the name of our function that does the plotting and a tuple for each parameter 
#representing (end_point0, end_point1, stepsize)
interact(plot, ω0 = (0,2,0.25), ω1 = (0,2,0.25));
````

## Exercise 08.3 (demographics and interactive plotting)

A county planning body has requested an interactive tool to visualise the population distribution in Cambridgeshire (by district) from 2021 to 2031 for different population growth rate scenarios in each district. It can be assumed that:

- the growth rates are constant in each district;
- the annual growth rate in any one district shall be between -10% and 10%.

Building on the pie chart example with population data in the Activity notebook, create an interactive plot with:

1. A slider for the year (from 2021 to 2031); and
2. Sliders for the annual population growth for each district (in percentage), with an 
   initial value of zero for each district.

### Solution

There are number of ways this exercise could be done, from very simple but not very robust with respect to changes in the order of the input data, to quite techical but robust with repsect to the order in which data is entered.

````python
# List of tuples containing district names and their 2021 population data
population_data = (('Cambridge City', 148500), 
                   ('East Cambridgeshire', 92630), 
                   ('Fenland', 107630), 
                   ('Huntingdonshire', 189440), 
                   ('South Cambridgeshire', 169800))
#extract a list of the district names
district = [district for district, population in population_data]
#Convert the population values into a 1D NumPy array for future mathematical calculations
population_initial = np.array([population for district, population in population_data])

#import numpy and matplotlib modules
import numpy as np
import matplotlib.pyplot as plt
#to create an interactive plot import the interactive widgets module ipywidgets
from ipywidgets import interact

#define an interactive plotting function that takes the %growth in each district as arguments to calculate future population growth
def plot (Cambridge_City = 0.0, East_Cambridgeshire = 0.0, Fenland = 0.0, Huntingdonshire = 0.0, South_Cambridgeshire = 0.0, year =2021):

    #store the inputted growth rates into a numpy array matching the order of the distrcits
    growth_rate = np.array([Cambridge_City, East_Cambridgeshire, Fenland, Huntingdonshire, South_Cambridgeshire])
    #calculate new population using compound growth formula
    population_new = population_initial * ((1 + (growth_rate /100)) ** (year - 2021))   

    
    #specify slice colours for pie chart
    colors = ['yellowgreen', 'gold', 'lightskyblue', 'lightcoral', 'red' ]
    #generate the pie chart using the new population data
    plt.pie(population_new, labels=district, colors=colors, autopct='%1.1f%%', shadow = True, startangle=90)
    plt.title(f'Population Distrubution in Cambridgeshire from 2021 to 2031') #add title
    plt.show()

#create interactive sliders which allows the user to adjust grwoth rates between -10% and 10% from 2021 - 2031
interact(plot, Cambridge_City=(-10.0, 10.0, 0.1), East_Cambridgeshire=(-10.0, 10.0, 0.1), Fenland=(-10.0, 10.0, 0.1), Huntingdonshire=(-10.0, 10.0, 0.1), South_Cambridgeshire=(-10.0, 10.0, 0.1), year = (2021, 2031, 1));
    
    
````

````python
#module imports

import json
import requests
import pprint
import matplotlib.pyplot as plt
import numpy as np 

#geographic coordinates for Jesus College, Cambridge
lat, long = 52.210306, 0.122734
#year-month of interest
year_month = '2025-03'

#construct the URL requst string
url = f"https://data.police.uk/api/crimes-street/all-crime?lat={lat}&lng={long}&date={year_month}"
#fetch data from https://data.ploice.uk
r = requests.get(url)
#check the data retrieval was succesful
r.raise_for_status()
#converts te fetched data into a list of python dictionaries
crime_data = r.json()

#can look at first report in the list, use pretty print module to make displayed data easier to read
if crime_data:
    pprint.pprint(crime_data[0])

#intialise an empty dictionary to keep track of crime category counts
categories_freq = {}

#iterate through each individual crime report in the retrieved dataset
for report in crime_data:
    #extract the crime category type identifier string from the report retrieved
    category = report['category']
    if category in categories_freq:
        #if the category is already a key in the tracking dictionary, increment its counter by 1
        categories_freq[category] += 1
    else:
        #if not, initialsie the category key in the dictionary with a count of 1
        categories_freq[category] = 1

#create a list of all the crime categories/ the dictionary keys
categories = list(categories_freq.keys())
#create a list of how many incidents recorded for each category/ the dictionary values
counts = list(categories_freq.values())

#create and array with equally spaced values with the position of each bar along the x axis
x_pos = np.arange(len(categories))
#produce the bar plot
plt.bar(x_pos, counts, align = 'center');
#replace the numeric index labels on the x axis with the crime category names
plt.xticks(x_pos, categories, rotation = 45, ha = 'right')

plt.xlabel('Reported Crimes') #label the x axis
plt.ylabel('No. of reports') #label the y axis
plt.show()


    
````

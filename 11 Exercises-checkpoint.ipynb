## Exercise 11.2

For the recursive factorial algorithm in Activity 04, determine the algorithmic complexity by inspecting your implementation of the algorithm. Test this against numerical experiments.

### Solution

Recall the factorial algorithm from Activity 04.4:

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return factorial(n - 1)*n
```

The function calls itself (recursively) $n$ times, hence it has complexity $O(n)$. We test this below and plot
the times.

```python
#create array of problem sizes N we want to test, generates numbers from 10 to 1000 with a step size of 50
N = np.arange(10, 1000, 50)

#initialise an empty array to store times for plotting
times = []

#time the search for each problem size
for n in N:
    #use magic command %timeit to measure execution time
    #-q: quiet mode (suppresses print output)
    #-n4: runs the loop 4 times per test
    #-r1: repeats the test cycle 1 time
    #-o: returns a TimeitResult object to access the best time
    t = %timeit -q -n4 -r1 -o factorial(int(n))
    #store the fastest execution time recorded for this value of n
    times.append(t.best)

#plot and label the log-log plot
plt.loglog(N, times, marker='o', label='factorial recursive')
plt.xlabel('$n$') #label x axis
plt.ylabel('$t$ (s)') #label y axis

#plotting a reference line for linear complexity
#multiplying n by a small constant factor aligns it close to the scale of the measured times
plt.loglog(N, 1e-7*N, label='$O(n)$ reference')
plt.legend(loc=0) #place the legend in the best location automatically
plt.show
```

## Exercise 11.3

Determine experimentally the complexity of computing the determinant of a matrix. You can generate an $n \times n$ matrix using:

```python
n = 100
A = np.random.rand(n, n)
```

and the determinant can be computed by:

```python
det = np.linalg.slogdet(A)
```

Be sure that you test for sufficiently large $n$ to get into the 'large' $n$ regime.

### Solution

Time computation of determinant:

```python
# Create array of problem sizes we want to test (powers of 2)
N = 2**np.arange(2, 12)
#initialise an empty array to store times for plotting
times = []
for n in N:
    #generate a random square matrix of size n*n 
    A = np.random.rand(n, n)
    #measure execution time of calculating determinant
    t = %timeit -q -n4 -r1 -o np.linalg.slogdet(A)
    #store fastest time to the list
    times.append(t.best)
```

Plot result:

```python
#plot matrix size vs measured times on a log-log graph
plt.loglog(N, times, marker='o', label='Determinant time')
#plot reference line showing O(n^3) growth
plt.loglog(N, 1e-10 * (N**3), label = '$O(n^3)$ reference')
plt.xlabel('Matrix Size $n$') #label x axis
plt.ylabel('Time $t$ (s)') #label y axis
plt.legend(loc = 0) #place the legend in the best place automatically
plt.show
```
